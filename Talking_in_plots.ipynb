```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd

path1 = "Case1_our_algo_3_arms.xlsx"
path2_a = "Case1_unconstrained_Lin_UCB_3_arms_reward.xlsx"
path2_b = "Case1_unconstrained_Lin_UCB_3_arms_utility.xlsx"
path3_a = "Case1_constrained_Lin_UCB_3_arms_reward_lambda_0.xlsx"
path3_b = "Case1_constrained_Lin_UCB_3_arms_utility_lambda_0.xlsx"

data1 = pd.read_excel(path1)
rew1,util1 = data1['reward'],data1['utility']
rew2 = pd.read_excel(path2_a)['reward']
rew3 = pd.read_excel(path3_a)['reward']
util2 = pd.read_excel(path2_b)['utility']
util3 = pd.read_excel(path3_b)['utility']
```

```python
rew1 = rew1.values.reshape(-1)
rew2 = rew2.values.reshape(-1)
rew3 = rew3.values.reshape(-1)

util1 = util1.values.reshape(-1)
util2 = util2.values.reshape(-1)
util3 = -util3.values.reshape(-1)
highest_reward = 0.96
optimal_reward = 0.7
highest_rew_util = 0.2
optimal_rew_util = 0.8
```

```python
plt.figure(figsize=(8,5))
plt.plot(np.cumsum(rew1)/np.arange(1,3001))
plt.plot(np.cumsum(rew2[0:3000])/np.arange(1,3001))
plt.plot(np.cumsum(rew3[0:3000])/np.arange(1,3001))
#plt.axhline(y=highest_reward,linestyle=':',color='brown')
plt.axhline(y=optimal_reward,linestyle='-.',color='red')
#plt.legend(["Ours","Unconstrained Lin_UCB","Primal Dual Lin_UCB",'Max mean reward','Optimal reward'])
plt.legend(["Ours","Unconstrained Lin_UCB","Primal Dual Lin_UCB",'Optimal reward'])
plt.xlabel("Iteration")
plt.ylabel("Average Reward")
plt.title("Avg Reward vs Iteration (3 arm) Lookup true reward and true utility")
plt.savefig('case1_3_arm_complete_average_reward.pdf', bbox_inches='tight')
plt.show()

plt.figure(figsize=(8,5))
plt.plot(np.cumsum(util1)/np.arange(1,3001))
plt.plot(np.cumsum(util2[0:3000])/np.arange(1,3001))
plt.plot(np.cumsum(util3[0:3000])/np.arange(1,3001))
plt.axhline(y=0.6,color='black',linestyle='--')
plt.axhspan(0.6, 1, facecolor='blue', alpha=0.225)
plt.axhspan(0, 0.6, facecolor='red', alpha=0.225)
#plt.axhline(y=highest_rew_util,linestyle=':',color='brown')
#plt.axhline(y=optimal_rew_util,linestyle='-.',color='red')
plt.legend(["Ours","Unconstrained Lin_UCB","Primal Dual Lin_UCB",'Boundary','Safe region','Unsafe region'])
#plt.legend(["Ours","Unconstrained Lin_UCB","Primal Dual Lin_UCB",'Boundary','Safe region','Unsafe region','Mean utility against max mean reward','Mean utility against optimal mean reward'])
plt.xlabel("Iteration")
plt.ylabel("Average Utility")
plt.title("Avg Reward vs Iteration (3 arm) Lookup true reward and true utility")
plt.savefig('case1_3_arm_complete_average_utility.pdf', bbox_inches='tight')
plt.show()
```
